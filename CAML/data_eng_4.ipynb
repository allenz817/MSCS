```python
import pandas as pd
titanic = pd.read_csv('titanic.csv')
```

```python
print(titanic.isnull())
```

```python
print(titanic.isnull().sum())
```

```python
df_dropped_rows = titanic.dropna(axis=0, how='any')
print(df_dropped_rows)
```

```python
df_dropped_columns = titanic.dropna(axis=1, how='any')
print(df_dropped_columns)

# arguments for how=any/all/<threadshold in number>
```
