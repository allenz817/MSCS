You have a dataset about ‘restaurant tips’, named tips.csv.  You are required to perform the analysis with python program using pandas on the following part, please include your code and result.   
1.	How many total bills are there for each day of the week?
2.	What is the maximum total bill amount for dinner?
3.	What is the average total bill amount for lunch and dinner?
4.	What is the highest tip given by a smoker?
5.	What is the average tip amount given by males and females?

```python
import pandas as pd
df = pd.read_csv("tips.csv")
```

```python
# 1.	How many total bills are there for each day of the week?
total_bills_by_day = df.groupby('day').size()
print(total_bills_by_day)
```

```python
# 2.	What is the maximum total bill amount for dinner?
max_total_bill_dinner = df[df['time'] == 'Dinner']['total_bill'].max()
print(max_total_bill_dinner)
```

```python
# 3.	What is the average total bill amount for lunch and dinner?
average_total_bill_lunch = df[df['time'] == 'Lunch']['total_bill'].mean()
average_total_bill_dinner = df[df['time'] == 'Dinner']['total_bill'].mean()
print(average_total_bill_lunch, average_total_bill_dinner)
```

```python
# 4.	What is the highest tip given by a smoker?
highest_tip_smoker = df[df['smoker'] == 'Yes']['tip'].max()
print(highest_tip_smoker)
```

```python
# 5.	What is the average tip amount given by males and females?
average_tip_male = df[df['sex'] == 'Male']['tip'].mean()
average_tip_female = df[df['sex'] == 'Female']['tip'].mean()
print(average_tip_male, average_tip_female)
```
