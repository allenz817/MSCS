```python
import pandas as pd
titanic = pd.read_csv('titanic.csv')
print(titanic.head(10))
```

```python
print(titanic.describe())
```

```python
print(titanic["Sex"].describe())
```

```python
print(titanic[5:9])
```

```python
max_age = titanic["Age"].max()
print(max_age)
```

```python
mean_age = titanic["Age"].mean()
print(mean_age)
```

```python
mean_survived = titanic["Survived"].mean()
print(mean_survived)
```

```python
ages = titanic["Age"]
print(ages.head(10))
```

```python
age_sex = titanic[["Age", "Sex"]] # a list of two columns, Age and Sex
print(age_sex.head(10))
```

```python
print(age_sex.notna().head(10)) # check if the values are not null
```

```python
print(age_sex.dropna().head(10)) # drop rows with null values
```

```python
# case: select specific rows based on a condition
above_35 = titanic[titanic["Age"] > 35]
print(above_35.head(10))
```

```python
female = titanic[titanic["Sex"] == "female"]
print(female.head(10))
```

```python
above_35_and_female = titanic[(titanic["Age"] > 35) & (titanic["Sex"] == "female")]
print(above_35_and_female.head(10))
```

```python
class_23 = titanic[titanic["Pclass"].isin([2, 3])]
print(class_23.head(10))
```

```python
class_23 = titanic[titanic["Pclass"] != 1]
print(class_23.head(10))
```

```python
# | is the logical OR operator, & is the logical AND operator
class_23 = titanic[(titanic["Pclass"] == 2) | (titanic["Pclass"] == 3)]
print(class_23.head(10).reset_index(drop=True))
# use reset_index(drop=True) to reset the index of the DataFrame after filtering
```

```python
age_not_na = titanic[titanic["Age"].notna()]
# use the .notna() method to filter out rows with null values in the Age column
# or use the .dropna() method to drop rows with null values in the Age column
# age_not_na = titanic.dropna(subset=["Age"])
print(age_not_na.head(10))
print(age_not_na.head(10))
```

```python
# complex data processing - select specific columns and rows
# using the .loc method
adult_names = titanic.loc[titanic["Age"] >= 18, ["Name", "Age"]]
print(adult_names.head(10))
```

```python
# using the .iloc method to select specific rows and columns by index
sample = titanic.iloc[9:25, 2:5]
print(sample)
```
